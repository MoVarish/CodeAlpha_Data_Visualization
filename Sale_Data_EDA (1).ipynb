```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'Product': ['Laptop', 'Mobile', 'Tablet',
                'Headphones', 'Keyboard', 'Mouse'],
    'Sales': [50000, 70000, 30000, 15000, 10000, 8000],
    'Units': [5, 10, 3, 15, 20, 25]
}

df = pd.DataFrame(data)

df
```

```python
print("First 5 Rows:")
print(df.head())

print("Rows and Columns:")
print(df.shape)

print("Dataset Information:")
print(df.info())

print("Statistical Summary:")
print(df.describe())
```

```python
print("Total Sales:", df['Sales'].sum())

print("Average Sales:", df['Sales'].mean())

print("Maximum Sales:", df['Sales'].max())

print("Minimum Sales:", df['Sales'].min())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("Duplicate Rows:")
print(df.duplicated().sum())
```

```python
plt.figure(figsize=(8,5))

sns.barplot(data=df, x='Product', y='Sales')

plt.title('Product Wise Sales Analysis')
plt.xlabel('Products')
plt.ylabel('Sales Amount')

plt.xticks(rotation=45)
plt.show()
```

```python
highest = df.loc[df['Sales'].idxmax()]
lowest = df.loc[df['Sales'].idxmin()]

print("Highest Sales Product:", highest['Product'])
print("Highest Sales Amount:", highest['Sales'])

print("Lowest Sales Product:", lowest['Product'])
print("Lowest Sales Amount:", lowest['Sales'])
```

```python
print("EXPLORATORY DATA ANALYSIS REPORT")

print("\n1. Dataset contains 6 rows and 3 columns.")

print("\n2. Total Sales: Rs.", df['Sales'].sum())

print("\n3. Average Sales: Rs.", df['Sales'].mean())

print("\n4. Highest Sales Product: Mobile")

print("\n5. Lowest Sales Product: Mouse")

print("\n6. Missing Values: None")

print("\n7. Duplicate Records: None")

print("\n8. Conclusion:")
print("Mobile has the highest sales, while Mouse has the lowest sales.")
```

Exploratory Data Analysis (EDA) – Observations

1. Mobile has the highest sales of ₹70,000.
2. Mouse has the lowest sales of ₹8,000.
3. Total sales are ₹1,83,000.
4. No missing values were found in the dataset.
5. No duplicate records were found in the dataset.

Conclusion

The analysis shows that Mobile is the best-selling product, while Mouse has the lowest sales. The dataset contains no missing or duplicate records.
