# Sales Data Analysis Dashboard
This notebook recreates the dashboard shown in the reference image using Python, Pandas, Matplotlib, and Seaborn.

**Charts included:** Product-wise sales, monthly sales trend, sales distribution, and sales/units heatmap.

```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Sample data transcribed from the reference dashboard
sales_data = pd.DataFrame({
    "Product": ["Laptop", "Mobile", "Tablet", "Headphones", "Keyboard", "Mouse"],
    "Sales":   [50000, 70000, 30000, 15000, 10000, 8000],
    "Units":   [5, 10, 3, 15, 20, 25]
})

monthly_sales = pd.DataFrame({
    "Month": ["January", "February", "March", "April", "May", "June"],
    "Sales": [50000, 60000, 45000, 75000, 90000, 85000]
})

display(sales_data)
display(monthly_sales)
```

```python
# Create the 2x2 Sales Analysis Dashboard
fig, axes = plt.subplots(2, 2, figsize=(16, 10))
fig.suptitle("SALES DATA ANALYSIS DASHBOARD", fontsize=19, fontweight="bold", y=0.98)

# 1. Product-wise Sales
axes[0, 0].bar(sales_data["Product"], sales_data["Sales"], color="#287db4")
axes[0, 0].set_title("Product Wise Sales")
axes[0, 0].set_xlabel("Products")
axes[0, 0].set_ylabel("Sales Amount")
axes[0, 0].tick_params(axis="x", rotation=30)

# 2. Monthly Sales Trend
axes[0, 1].plot(monthly_sales["Month"], monthly_sales["Sales"],
                marker="o", color="#1678c8")
axes[0, 1].set_title("Monthly Sales Trend")
axes[0, 1].set_xlabel("Months")
axes[0, 1].set_ylabel("Sales")
axes[0, 1].tick_params(axis="x", rotation=30)
axes[0, 1].grid(True, alpha=0.7)

# 3. Sales Distribution
axes[1, 0].pie(
    sales_data["Sales"],
    labels=sales_data["Product"],
    autopct="%1.1f%%",
    startangle=0
)
axes[1, 0].set_title("Sales Distribution")
axes[1, 0].axis("equal")

# 4. Sales and Units Heatmap
heatmap_data = sales_data[["Sales", "Units"]]
sns.heatmap(
    heatmap_data,
    annot=True,
    fmt="g",
    cmap="YlOrRd",
    cbar=True,
    ax=axes[1, 1]
)
axes[1, 1].set_title("Sales and Units Heatmap")
axes[1, 1].set_xlabel("")
axes[1, 1].set_ylabel("")

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

```python
# Optional: save the dashboard as a PNG image
fig, axes = plt.subplots(2, 2, figsize=(16, 10))
fig.suptitle("SALES DATA ANALYSIS DASHBOARD", fontsize=19, fontweight="bold", y=0.98)

axes[0, 0].bar(sales_data["Product"], sales_data["Sales"], color="#287db4")
axes[0, 0].set_title("Product Wise Sales")
axes[0, 0].set_xlabel("Products")
axes[0, 0].set_ylabel("Sales Amount")
axes[0, 0].tick_params(axis="x", rotation=30)

axes[0, 1].plot(monthly_sales["Month"], monthly_sales["Sales"], marker="o", color="#1678c8")
axes[0, 1].set_title("Monthly Sales Trend")
axes[0, 1].set_xlabel("Months")
axes[0, 1].set_ylabel("Sales")
axes[0, 1].tick_params(axis="x", rotation=30)
axes[0, 1].grid(True, alpha=0.7)

axes[1, 0].pie(sales_data["Sales"], labels=sales_data["Product"], autopct="%1.1f%%", startangle=0)
axes[1, 0].set_title("Sales Distribution")
axes[1, 0].axis("equal")

sns.heatmap(sales_data[["Sales", "Units"]], annot=True, fmt="g", cmap="YlOrRd", cbar=True, ax=axes[1, 1])
axes[1, 1].set_title("Sales and Units Heatmap")
axes[1, 1].set_xlabel("")
axes[1, 1].set_ylabel("")

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.savefig("sales_data_analysis_dashboard.png", dpi=200, bbox_inches="tight")
plt.show()
print("Dashboard saved as sales_data_analysis_dashboard.png")
```
